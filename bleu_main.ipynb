```python
import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction

# Utility to read and tokenize a file
def read_and_tokenize(filepath):
    with open(filepath, 'r') as f:
        text = f.read()
    return text.strip().split()

# Paths to your files (update as needed)
file1 = "peac_modules_single/1_standup_comedy.txt"
file2 = "peac_modules_single/1_standup_comedy_peac.txt"

# Read and tokenize
reference = [read_and_tokenize(file1)]
candidate = read_and_tokenize(file2)

# Compute BLEU with smoothing for short texts
smooth = SmoothingFunction().method1
bleu_score = sentence_bleu(reference, candidate, smoothing_function=smooth)

print(f"BLEU score {bleu_score:.4f}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Read files as plain text
def read_text(filepath):
    with open(filepath, 'r') as f:
        return f.read()

file1 = "peac_modules_single/2_commentariat.txt"

text1 = read_text(file1)
text2 = read_text(file2)

# Vectorize and compute cosine similarity
vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform([text1, text2])
similarity = cosine_similarity(tfidf[0:1], tfidf[1:2])[0][0]

print(f"Cosine similarity between {file1} and {file2}: {similarity:.4f}")
```

```python
from typing import Dict, List


class SentenceRecord:
    def __init__(self, file: str, section: str, sentence: str):
        self.file = file
        self.section = section
        self.sentence = sentence

    def as_dict(self) -> Dict:
        return {
            "file": self.file,
            "section": self.section,
            "sentence": self.sentence
        }

    def __repr__(self):
        return f"SentenceRecord(file={self.file}, section={self.section}, sentence={self.sentence})"

class PeacSentenceExtractor:
    def __init__(self, yaml_path: str):
        self.yaml_path = yaml_path
        self.py = PromptYaml(yaml_path, '', False)

    def extract(self) -> List[SentenceRecord]:
        records = []
        # Instruction
        for s in self.py.get_instruction_base_rules():
            if s and s.strip():
                records.append(SentenceRecord(self.yaml_path, "instruction", s))
        # Context
        for s in self.py.get_context_base_rules():
            if s and s.strip():
                records.append(SentenceRecord(self.yaml_path, "context", s))
        # Output
        for s in self.py.get_output_base_rules():
            if s and s.strip():
                records.append(SentenceRecord(self.yaml_path, "output", s))
        # Query
        q = self.py.get_query()
        if q and isinstance(q, str) and q.strip():
            records.append(SentenceRecord(self.yaml_path, "query", q))
        return records
```

```python
def get_yaml():
    import os
    directory = "peac_modules_single"
    return [os.path.join(directory, f) for f in os.listdir(directory) if f.endswith('.yaml') and not 'peac' in f]

from sentence_transformers import SentenceTransformer
from sklearn.cluster import AgglomerativeClustering
import csv

yamls = get_yaml()
# collect all sentences from all yamls
def collect_all_sentences(yaml_files: List[str]) -> List[SentenceRecord]:
    all_records = []
    for yf in yaml_files:
        print(f"Processing {yf}")
        extractor = PeacSentenceExtractor(yf)
        records = extractor.extract()
        all_records.extend(records)
    return all_records
def cluster_sentences(records: List[SentenceRecord], threshold=0.7):
    if not records:
        return []
    model = SentenceTransformer('all-MiniLM-L6-v2')
    sentences = [r.sentence for r in records]
    embeddings = model.encode(sentences)
    clustering = AgglomerativeClustering(
        n_clusters=None,
        metric='cosine',
        linkage='average',
        distance_threshold=1-threshold
    )
    labels = clustering.fit_predict(embeddings)
    return labels

def save_clusters_to_csv(records: List[SentenceRecord], labels, out_csv: str):
    with open(out_csv, 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(['cluster_id', 'file', 'section', 'sentence'])
        for record, label in zip(records, labels):
            writer.writerow([label, record.file, record.section, record.sentence])
```
